# Dropout schedules in a ReLU MLP

If dropout shortens the depth over which correlations survive, its placement should matter as well as its strength. We test that idea on CIFAR-10 by moving dropout across the hidden layers of the same ReLU MLP, comparing uniform, early, and late schedules with a no-dropout control. The initialization uses $\sigma_w^2=1.98$ and $\sigma_b^2=0.02$, and the depth is chosen from the near-critical correlation-length proxy as $L=\lfloor2\xi_{\mathrm{constant}}\rfloor$.

To inspect the saved runs, execute **Setup** and **Correlation length**, then jump to **Saved results** and **Figures**. Select `mlp_budget_controls.npz` in the loading cell to inspect the budget controls; running the full notebook trains new models.

```python
from utils import get_dropout_schedule

# Theory functions


def compute_effective_xi(h_layers, schedule_type=None, h_bar_target=None):
    """Compute effective correlation length (kinked/ReLU class)."""
    kappa = 2 * np.sqrt(2) / (3 * np.pi)
    alpha = (3 / 2) * kappa ** (2 / 3)
    L = len(h_layers)

    damage = sum(h ** (1 / 3) for h in h_layers if h > 0) / L
    h_bar_actual = np.mean(h_layers)
    h_max_actual = max(h_layers) if h_layers else 0
    n_dropout = sum(1 for h in h_layers if h > 0)

    xi_inv = alpha * damage
    xi_eff = 1 / xi_inv if xi_inv > 0 else float("inf")

    return {
        "xi_eff": xi_eff,
        "h_bar": h_bar_actual,
        "h_max": h_max_actual,
        "n_dropout": n_dropout,
    }
```

## Correlation length

The code retains the names `H_BAR` and `h_layers` for the raw dropout probabilities $p_\ell$ passed to PyTorch. We use them as proxies for the theoretical field $h=1-F(1)$ when setting the depth and comparing schedules; the two quantities are not generally equal. The resulting $\xi$ estimates describe a near-critical mean-field surrogate, whose relevance to these finite, trained networks is what the experiments test.


```python
# Compute depth from correlation length

# Correlation length for constant dropout (kinked class)
kappa = 2 * np.sqrt(2) / (3 * np.pi)
alpha = (3 / 2) * kappa ** (2 / 3)
xi_constant = 1 / (alpha * H_BAR ** (1 / 3))

# Depth as multiple of correlation length
DEPTH = int(DEPTH_MULTIPLIER * xi_constant)

print("=" * 60)
print("CONFIGURATION")
print("=" * 60)
print(f"σ_w² = {SIGMA_W_SQ:.4f}  (He init = 2.0)")
print(f"σ_b² = {SIGMA_B_SQ:.4f}")
print(f"h̄ = {H_BAR:.4f}  (dropout field)")
print("")
print(f"Correlation length (constant): ξ = {xi_constant:.2f}")
print(f"Network depth: L = {DEPTH}  ({DEPTH_MULTIPLIER}× ξ)")
print(f"Ratio L/ξ = {DEPTH / xi_constant:.2f}")
print("")
print("Regime check:")
print(f"  L/N = {DEPTH / WIDTH:.4f}")
print(f"  h̄/(L/N) = {H_BAR / (DEPTH / WIDTH):.1f}")
status = "✓ dropout dominates" if H_BAR > DEPTH / WIDTH else "⚠ non-Gaussianities may matter"
print(f"  {status}")
print("=" * 60)
```

```python
# Theory predictions

theory = {}
print(f"{'Schedule':<15} {'h̄':>8} {'h_max':>8} {'#drop':>6} {'ξ_eff':>8} {'L/ξ':>6}")
print("-" * 60)

for sched in SCHEDULES:
    h_layers = get_dropout_schedule(sched, DEPTH, H_BAR, H_MAX)
    stats = compute_effective_xi(h_layers, schedule_type=sched, h_bar_target=H_BAR)
    theory[sched] = {"h_layers": h_layers, **stats}

    xi_str = f"{stats['xi_eff']:.2f}" if stats["xi_eff"] < 1e6 else "∞"
    ratio = DEPTH / stats["xi_eff"] if stats["xi_eff"] < 1e6 else 0
    n_drop_str = (
        f"{stats['n_dropout']:.1f}"
        if isinstance(stats["n_dropout"], float)
        else f"{stats['n_dropout']}"
    )
    print(
        f"{sched:<15} {stats['h_bar']:>8.4f} {stats['h_max']:>8.4f} "
        f"{n_drop_str:>6} {xi_str:>8} {ratio:>6.2f}"
    )
```

## Training

This section downloads CIFAR-10 and trains the models with the configuration above. New results and figures are written to `runs/mlp_schedules/`.


```python
# Model with σ_b² > 0


class CriticalReLUNet(nn.Module):
    def __init__(
        self,
        input_dim,
        hidden_dim,
        output_dim,
        h_layers,
        sigma_w_sq=2.0,
        sigma_b_sq=0.0,
        schedule_type=None,
        h_bar=None,
        seed=None,
    ):
        super().__init__()
        self.depth = len(h_layers)
        self.sigma_w_sq = sigma_w_sq
        self.sigma_b_sq = sigma_b_sq
        self.schedule_type = schedule_type
        self.h_bar = h_bar
        self.seed = seed

        layers = [nn.Linear(input_dim, hidden_dim)]
        for _ in range(self.depth - 1):
            layers.append(nn.Linear(hidden_dim, hidden_dim))
        layers.append(nn.Linear(hidden_dim, output_dim))

        self.layers = nn.ModuleList(layers)
        self.dropouts = nn.ModuleList([nn.Dropout(p=h) for h in h_layers])
        self._init_critical()

    def _init_critical(self):
        for layer in self.layers:
            fan_in = layer.weight.shape[1]
            std_w = np.sqrt(self.sigma_w_sq / fan_in)
            nn.init.normal_(layer.weight, mean=0.0, std=std_w)

            if self.sigma_b_sq > 0:
                nn.init.normal_(layer.bias, mean=0.0, std=np.sqrt(self.sigma_b_sq))
            else:
                nn.init.zeros_(layer.bias)

    def forward(self, x):
        x = x.view(x.size(0), -1)
        for i, layer in enumerate(self.layers[:-1]):
            x = torch.relu(layer(x))
            x = self.dropouts[i](x)
        return self.layers[-1](x)
```

```python
# Training utilities (from shared utils)

from utils import train_epoch, evaluate, load_cifar
```

```python
# Load data

print("Loading CIFAR-10...")
train_data, test_data = load_cifar(10, device, TRAIN_SIZE, TEST_SIZE)
print(f"Train: {train_data[0].shape}, Test: {test_data[0].shape}")
```

```python
# Run experiments


def run_experiments(schedules, n_sims, epochs):
    results = {}

    for sched in schedules:
        print(f"\n{'=' * 50}")
        print(f"Schedule: {sched}")
        print(f"{'=' * 50}")

        h_layers = theory[sched]["h_layers"]
        histories = []

        for sim in range(n_sims):
            seed = 42 + sim
            torch.manual_seed(seed)
            np.random.seed(seed)

            model = CriticalReLUNet(
                input_dim=3072,
                hidden_dim=WIDTH,
                output_dim=10,
                h_layers=h_layers,
                sigma_w_sq=SIGMA_W_SQ,
                sigma_b_sq=SIGMA_B_SQ,
                schedule_type=sched,
                h_bar=H_BAR,
                seed=seed,
            ).to(device)

            opt = optim.AdamW(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-7)
            scheduler = (
                CosineAnnealingLR(opt, T_max=epochs, eta_min=LR_MIN) if USE_LR_SCHEDULE else None
            )
            crit = nn.CrossEntropyLoss()

            hist = {"train_acc": [], "test_acc": [], "train_loss": [], "test_loss": []}
            best_test_acc = 0.0
            best_test_loss = float("inf")

            pbar = tqdm(range(epochs), desc=f"{sched} sim {sim + 1}/{n_sims}", leave=True)

            for ep in pbar:
                tr_loss, tr_acc = train_epoch(
                    model, train_data, opt, crit, BATCH_SIZE, amp_dtype=AMP_DTYPE
                )
                if scheduler:
                    scheduler.step()

                if ep % EVAL_EVERY == 0 or ep == epochs - 1:
                    te_loss, te_acc = evaluate(
                        model, test_data, crit, BATCH_SIZE, amp_dtype=AMP_DTYPE
                    )
                else:
                    te_loss, te_acc = (
                        hist["test_loss"][-1] if hist["test_loss"] else 0,
                        hist["test_acc"][-1] if hist["test_acc"] else 0,
                    )

                # Track best
                if te_acc > best_test_acc:
                    best_test_acc = te_acc
                if te_loss < best_test_loss:
                    best_test_loss = te_loss

                hist["train_acc"].append(tr_acc)
                hist["test_acc"].append(te_acc)
                hist["train_loss"].append(tr_loss)
                hist["test_loss"].append(te_loss)

                # Update progress bar
                pbar.set_postfix(
                    {
                        "ep": f"{ep + 1}/{epochs}",
                        "tr_L": f"{tr_loss:.3f}",
                        "tr_A": f"{tr_acc:.1f}%",
                        "te_L": f"{te_loss:.3f}",
                        "te_A": f"{te_acc:.1f}%",
                        "best": f"{best_test_acc:.1f}%",
                    }
                )

            histories.append(hist)

        # Aggregate
        results[sched] = {
            "train_acc": np.array([h["train_acc"] for h in histories]),
            "test_acc": np.array([h["test_acc"] for h in histories]),
            "train_loss": np.array([h["train_loss"] for h in histories]),
            "test_loss": np.array([h["test_loss"] for h in histories]),
        }

        best = results[sched]["test_acc"].max(axis=1)
        print(f"{sched}: Best test = {best.mean():.2f}% ± {best.std():.2f}% (SD)")

    return results


results = run_experiments(SCHEDULES, N_SIMULATIONS, EPOCHS)
```

```python
# Single comprehensive figure for paper

fig = plt.figure(figsize=(7, 7))

# Create grid spec for custom layout
gs = fig.add_gridspec(3, 2, height_ratios=[1, 1, 0.9], hspace=0.35, wspace=0.3)

ax_loss_train = fig.add_subplot(gs[0, 0])
ax_loss_test = fig.add_subplot(gs[0, 1])
ax_acc_train = fig.add_subplot(gs[1, 0])
ax_acc_test = fig.add_subplot(gs[1, 1])
ax_dist = fig.add_subplot(gs[2, :])

# Plot training curves
for sched in SCHEDULES:
    c = get_color(sched)
    xi = theory[sched]["xi_eff"]
    xi_str = f"{xi:.1f}" if xi < 1e6 else r"$\infty$"
    label = f"{get_display_name(sched)} ($\\xi$={xi_str})"

    # Training loss
    m = results[sched]["train_loss"].mean(0)[5:]
    s = results[sched]["train_loss"].std(0)[5:]
    epochs_plot = np.arange(5, 5 + len(m))
    ax_loss_train.semilogy(epochs_plot, m, color=c, lw=1.5)
    ax_loss_train.fill_between(epochs_plot, np.maximum(m - s, 1e-6), m + s, color=c, alpha=0.15)

    # Test loss
    m = results[sched]["test_loss"].mean(0)[5:]
    s = results[sched]["test_loss"].std(0)[5:]
    ax_loss_test.semilogy(epochs_plot, m, color=c, lw=1.5)
    ax_loss_test.fill_between(epochs_plot, np.maximum(m - s, 1e-6), m + s, color=c, alpha=0.15)

    # Training accuracy
    m = results[sched]["train_acc"].mean(0)
    s = results[sched]["train_acc"].std(0)
    epochs_full = np.arange(len(m))
    ax_acc_train.plot(epochs_full, m, color=c, lw=1.5)
    ax_acc_train.fill_between(epochs_full, m - s, m + s, color=c, alpha=0.15)

    # Test accuracy
    m = results[sched]["test_acc"].mean(0)
    s = results[sched]["test_acc"].std(0)
    ax_acc_test.plot(epochs_full, m, color=c, label=label, lw=1.5)
    ax_acc_test.fill_between(epochs_full, m - s, m + s, color=c, alpha=0.15)

# Configure training curve axes
ax_loss_train.set_xlabel("Epoch")
ax_loss_train.set_ylabel("Training loss")
ax_loss_train.set_title("(a) Training loss", fontweight="normal", loc="left")

ax_loss_test.set_xlabel("Epoch")
ax_loss_test.set_ylabel("Test loss")
ax_loss_test.set_title("(b) Test loss", fontweight="normal", loc="left")

ax_acc_train.set_xlabel("Epoch")
ax_acc_train.set_ylabel("Training accuracy (%)")
ax_acc_train.set_title("(c) Training accuracy", fontweight="normal", loc="left")

ax_acc_test.set_xlabel("Epoch")
ax_acc_test.set_ylabel("Test accuracy (%)")
ax_acc_test.set_title("(d) Test accuracy", fontweight="normal", loc="left")
ax_acc_test.legend(loc="lower right", fontsize=7, framealpha=0.95)

# Bottom panel: Box plot
box_data = [final_accs[sched] for sched in SCHEDULES]
box_labels = [get_display_name(sched) for sched in SCHEDULES]
bp = ax_dist.boxplot(box_data, patch_artist=True, widths=0.5, vert=True)
ax_dist.set_xticks(range(1, len(box_labels) + 1), box_labels)

for patch, sched in zip(bp["boxes"], SCHEDULES):
    patch.set_facecolor(get_color(sched))
    patch.set_alpha(0.6)
    patch.set_edgecolor("0.3")
    patch.set_linewidth(0.8)

for element in ["whiskers", "caps", "medians"]:
    for item in bp[element]:
        item.set_color("0.3")
        item.set_linewidth(0.8)

ax_dist.set_ylabel("Final test accuracy (%)")
ax_dist.set_title("(e) Final accuracy distribution", fontweight="normal", loc="left")
ax_dist.tick_params(axis="x", rotation=20)

plt.savefig(RUN_DIR / "comprehensive_figure.pdf", format="pdf")
plt.savefig(RUN_DIR / "comprehensive_figure.png", dpi=300)
plt.show()
```
